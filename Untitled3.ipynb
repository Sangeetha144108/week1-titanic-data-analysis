```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully!")
print(df.shape)
```

```python
survival_gender = df.groupby('Sex')['Survived'].mean() * 100

plt.figure(figsize=(8, 5))
ax = sns.barplot(
    x=survival_gender.index,
    y=survival_gender.values
)

plt.title('Titanic Survival Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Survival Rate (%)')
plt.ylim(0, 100)

for i, value in enumerate(survival_gender.values):
    ax.text(i, value + 2, f'{value:.1f}%', ha='center')

plt.tight_layout()
plt.show()
```

```python
survival_class = df.groupby('Pclass')['Survived'].mean() * 100

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x=survival_class.index,
    y=survival_class.values
)

plt.title('Titanic Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate (%)')
plt.ylim(0, 100)

for i, value in enumerate(survival_class.values):
    ax.text(i, value + 2, f'{value:.1f}%', ha='center')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='Age',
    hue='Survived',
    bins=20,
    kde=True
)

plt.title('Age Distribution of Titanic Passengers by Survival')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.tight_layout()
plt.show()
```

```python
# Visualization 4: Survival Rate by Gender and Passenger Class

survival_gender_class = (
    df.groupby(['Pclass', 'Sex'])['Survived']
    .mean()
    .reset_index()
)

survival_gender_class['Survival Rate (%)'] = (
    survival_gender_class['Survived'] * 100
)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=survival_gender_class,
    x='Pclass',
    y='Survival Rate (%)',
    hue='Sex'
)

plt.title('Titanic Survival Rate by Gender and Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate (%)')
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
numeric_data = df.select_dtypes(include=['number'])

# Create correlation matrix
corr = numeric_data.corr()

# Plot heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```
